```python
import matplotlib.pyplot as plt
import numpy as np
men=[22,30,35,35,26]
women=[25,32,30,35,29]
x_axis=np.arange(len(men))
plt.bar(x_axis-0.2,men,width=0.3,color='blue',label='Men')
plt.bar(x_axis+0.2,women,width=0.3,color='green',label='Women')
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Bar Plot of Men and Women Scores")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
Unemployment_Rate=(6.1,5.8,5.7,5.7,5.8,5.6,5.5,5.3,5.2,5.2)
Stock_index_Price=[1500,1520,1525,1523,1515,1540,1545,1560,1555,1565]
plt.scatter(Unemployment_Rate,Stock_index_Price)
plt.title('Scatter plot-Stock_index_Price vs Unemployment_Rate',color='green')
plt.xlabel('Unemployment_Rate',fontsize=14)
plt.ylabel('Stock_index_Price',fontsize=14)
plt.xlabel('Unemployment_Rate')
plt.ylabel('Stock_index_Price')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
value1=[72,76,24,40,57,67,75,78,31,32]
value2=[62,5,91,25,36,32,96,95,30,90]
value3=[23,89,12,78,72,89,25,69,68,86]
value4=[99,73,70,16,81,61,88,98,10,87]
box_plot_data=[value1,value2,value3,value4]
box=plt.boxplot(box_plot_data,vert=0,patch_artist=True,labels=['course1','course2','course3','course4'])
colors=['cyan','lightgreen','tan']
for patch,color in zip(box['boxes'],colors):
  patch.set_facecolor(color)
plt.show()
```

```python
import matplotlib. pyplot as plt
une_rate=(6.1,5.8,5.7,5.7,5.8,5.6,5.5,5.3,5.2,5.2)
stk=[1500,1520,1525,1523,1515,1540,1545,1560,1555,1565]
plt.scatter(une_rate,stk,color='green')
plt.title('Unemployment Rate vs stock index price',fontsize=14)
plt.xlabel('Unemployment Rate',fontsize=14)
plt.ylabel('stock index price',fontsize=14)
plt.grid(True)
plt.show()
```
