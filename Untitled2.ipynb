```python
import matplotlib.pyplot as plt
x=[-2,-1,0,1,2]
y=[i**2for i in x]
plt.plot(x,y)
plt.title("Line Graph-y=x^2",fontsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
weights=[78,72,69,81,63,66,65,79,74,71,79,80]
plt.hist(weights,
         bins=[50,55,60,65,70,75,80,85],
               color="red",
               edgecolor="blue")
plt.title("Histogram of Muffin weight")
plt.xlabel("Weight in grms")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
class1=['I','II','III','IV','V','VI','VII','VIII','IX','X']
strength=[40,43,45,47,49,38,50,37,43,39]
colors=['red','yellow','green',
        'blue','maroon','orange','violet',
        'grey','cyan','black']
plt.bar(class1,strength,color=colors,align='center')
plt.xlabel("class")
plt.ylabel("Strength")
plt.show()
```
