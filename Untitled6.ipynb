```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
data={'BMI':[33.5,28.6,23.4,43.1,35.3,35.9,36.7,25.7,23.3,31],
'AGE':[50,30,40,67,23,67,45,46,29,56],
'SUGAR':[1,0,0,0,1,1,1,0,0,1]}
df=pd.DataFrame(data)
print("Dataset:")
print(df)
plt.scatter(
    df['AGE'],df['BMI'],c=df['SUGAR'].map({0:'red',1:'blue'}),s=100
)
plt.plot([],[],color='red',label='Sugar is 0')
plt.plot([],[],color='blue',label='Sugar is 1')
plt.legend(title='Sugar level')
plt.title('Scatter Plot of Age vs BMI')
plt.xlabel('Age')
plt.ylabel('BMI') # Added y-axis label for clarity
plt.grid(True) # Corrected 'BMI' to True for grid display
plt.show()
x=df[['BMI','AGE']]
y=df['SUGAR']
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.5,random_state=42)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
k=3
knn=KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train_scaled,y_train) # Fit with scaled data
y_pred=knn.predict(X_test_scaled) # Predict with scaled data
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy:",accuracy)
print("\nClassification Report:")
print(classification_report(y_test,y_pred,zero_division=0))
cm=confusion_matrix(y_test,y_pred)
print("\nConfusion Matrix:")
print(cm)
new_data=pd.DataFrame([[43.6,40]],columns=['BMI','AGE'])
new_data_scaled=scaler.transform(new_data)
pred=knn.predict(new_data_scaled)
print("\nnew data:")
print("BMI=43.6")
print("AGE=40")
print ("Predicted.",pred[0])
if pred[0]==0:
 print("The person has no sugar")
else:
 print("The person has sugar")
```
