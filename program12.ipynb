```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
from sklearn.datasets import fetch_california_housing
housing=fetch_california_housing(as_frame=True)
df=housing.frame
print("First 5 records")
print(df.head())
print("\n Dataset shape")
print(df.shape)
x=df[['MedInc']]
y=df['MedHouseVal']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
linear_model=LinearRegression()
linear_model.fit(x_train,y_train)
y_pred=linear_model.predict(x_test)
mse_linear=mean_squared_error(y_test,y_pred)
rmse_linear=np.sqrt(mse_linear)
r2_linear=r2_score(y_test,y_pred)
print("\n Linear Regression ")
print("Coefficient:",linear_model.coef_[0])
print("Intercept",linear_model.intercept_)
print(" MSE:",mse_linear)
print("RMSE:",rmse_linear)
print("R2score:",r2_linear)
x=df.drop('MedHouseVal',axis=1)
y=df['MedHouseVal']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
multiple_model=LinearRegression()
multiple_model.fit(x_train,y_train)
y_pred_multiple=multiple_model.predict(x_test)
mse_multiple=mean_squared_error(y_test,y_pred_multiple)
rmse_multiple=np.sqrt(mse_multiple)
r2_multiple=r2_score(y_test,y_pred_multiple)
print("\n Multiple Regression")
print("Coefficients:",multiple_model.coef_)
print("Intercept:",multiple_model.intercept_)
print("MSE:",mse_multiple)
print("RMSE:",rmse_multiple)
print("R2score:",r2_multiple)
print("\n Performance Comparison")
comparison=pd.DataFrame({'Model':['Linear','Multiple'],'MSE':[mse_linear,mse_multiple],'RMSE':[rmse_linear,rmse_multiple],'R2score':[r2_linear,r2_multiple]})
print(comparison)
plt.scatter(x_test['MedInc'],y_test,label='Actual')
plt.scatter(x_test['MedInc'],y_pred_multiple,label='Predicted')
plt.xlabel('Media Income')
plt.ylabel('Median House Value')
plt.title('linear Regression')
plt.legend()
plt.show()
```
